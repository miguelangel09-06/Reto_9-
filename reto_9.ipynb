1.Desarrollar un algoritmo que calcule el promedio de un arreglo de reales.

```python
numeros_reales=[7,8,12,560,8.35,754,98,97]
cantidad=len(numeros_reales)
suma=0
for i in numeros_reales:
  suma+=i
promedio=suma/cantidad 
print("El promedio de la lista es =", promedio)
```

2.Desarrollar un algoritmo que calcule el producto punto de dos arreglos de números enteros (reales) de igual tamaño.

```python
lista1=[2,5,2,40,17,14,23,64,71,39]
lista2=[3,11,8,7,5,22,47,33,34,23]
lleva_suma=0
for i in range(len(lista1)):
    lleva_suma+=lista1[i]*lista2[i]

print("El producto punto de los arreglos es =",lleva_suma)
```

3.Hacer un algoritmo que deje al final de un arreglo de números todos los ceros que aparezcan en dicho arreglo.

```python

#Se crean dos listas, una que lleva los datos originales y la otra donde entran los datos ordenados
arreglo_numeros=[2,10,0,3,8,8,75,9,7,5,0,5,0,45,0,39]
lista_repuesto=[]
#Este for recorre la lista y cuando un numero es diferente de cero lo agerga a la de repuesto primero
for i in arreglo_numeros:
    if i!=0:
        lista_repuesto.append(i)
#Este for recorre la lista y cuando un numero es igual a cero lo agerga a la de repuesto de ultimo
for j in arreglo_numeros:
    if j==0:
        lista_repuesto.append(j)
#Se imprime       
print("Se imprime la lista normal=",arreglo_numeros)
print("Se imprime la lista con los ceros al final=",lista_repuesto)

#TAmbien se puede usar la funcion .sort que ordena los numeros de menor a mayor de derecha a izquierda 
arreglo_numeros.sort(reverse=True)
print("Se imprime la lista con los ceros al final con otro metodo=",arreglo_numeros)
```
